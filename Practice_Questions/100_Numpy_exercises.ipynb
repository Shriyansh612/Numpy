#### 46. Create a structured array with `x` and `y` coordinates covering the [0,1]x[0,1] area (★★☆)

````python
coordinate = np.dtype([('x',float),('y',float)])

x = np.linspace(0,1,6,endpoint=True)
y = np.linspace(0,1,6,endpoint=True)

space = np.zeros((6,6),dtype=coordinate)
X,Y = np.meshgrid(x,y,indexing = 'xy')

space["x"] = X
space["y"] = Y
print(space)
````
