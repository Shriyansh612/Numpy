####    Question 3
        Create a 1D array of 5 equally spaced numbers between 0.0 and 1.0 using np.linspace()

```python
arr = np.linspace(0,1,num=5)
print(arr)
```

#### SYNTAX

numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)

• start: The starting value of the sequence.

• stop: The end value of the sequence.

• num (Optional): The number of samples to generate. The default is 50.

• endpoint (Optional): If True (default), the stop value is included in the array. If False, the stop value is excluded.

• retstep (Optional): If set to True, the function returns a tuple containing (array, step_size). The default is False.

• dtype (Optional): The data type of the output array. If not specified, it defaults to a floating-point type.

#### Question 12
    Given arr = np.array([5, -2, 10, -8, 3]), create a boolean mask for values strictly greater than 0

```python
# Boolean Indexing
arr = np.array([5,-2,10,-8,3])
positive_mask = arr>0   #boolean mask for positive numbers
print(positive_mask)

# Filtering using Boolean Mask
positive = arr[positive_mask]# boolean indexing   #values in arr satisfying the positive mask
print(positive)
```

#### Question 13
    Combining two conditions
    


```python
arr = np.array([-4,-9,2,4,0,9,-6])
combined_condition = (arr>0) & (arr<5)  #bitwise and paranthesis are required because of precedence
print(arr[combined_condition])
```

To combine multiple criteria, you must use bitwise operators (& for AND, | for OR, ~ for NOT). Standard Python operators (and, or) will throw an error.

⚠️ Crucial Rule: You must wrap each condition in parentheses, or the order of operations will fail.

#### Question 15
    To get the indices of elements satisfying a specific condition
    Find indices of elements greater than 20

```python
arr = np.array([40,10,32,6,15,155,85])
print(np.where(arr>20))
print(np.where(arr>20)[0])
```

When you pass only a condition to np.where(), it returns a tuple of arrays (one for each dimension) containing the indices where the condition evaluates to True

Or use np.nonzero(condition) similarly



#### Question 16

#### Question 22
    Remove a single-dimensional entry from an array of shape (1, 5) using np.squeeze()

```python
arr = np.array([[1,2,3,4,5]])
print(arr,arr.size)
arr1 = arr.squeeze()
print(arr1,arr1.size)
arr2 = arr.reshape(5)
print(arr2,arr2.size)
```
