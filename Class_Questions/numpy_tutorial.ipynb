```python
import numpy as np
import time
```

```python
# Execution performance
# py_list = [1, 2, 3, 4, 5]

# Python list
size = 10000
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list time : {end-start}seconds")



# numpy arrays
np_arr = np.array(py_list)  
start = time.time()
# vectorizetion
sq_array = np_arr ** 2
end = time.time()
print(f"Numpy array time : {end-start}seconds")
```

```python
import sys

print(f"Python list size = {sys.getsizeof(py_list)} bytes ")
print(f"numpy arrays size = {np_arr.nbytes} bytes ")
```
