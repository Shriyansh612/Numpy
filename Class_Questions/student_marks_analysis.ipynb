```python
import numpy as np
```

```python
marks = np.array([[79,67,60],
                  [59,92,59],
                  [39,60,95],
                  [87,83,76]])

total_marks = np.sum(marks,axis=1)
print(total_marks,type(total_marks))

mean_marks = np.mean(marks,axis=1)
mean_marks = mean_marks.round(2)
print(mean_marks)

highest_in_each_subject = np.max(marks,axis=0)
print(highest_in_each_subject)

# l = mean_marks.tolist()
# print(l,type(l))

student_highest_marks = np.argmax(total_marks)
print(student_highest_marks)

freq_marks = np.sum(marks>=80)
print(freq_marks)

freq_marks2 = np.sum(marks>=80,axis=1)
print(freq_marks2)
```
