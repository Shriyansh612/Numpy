```python
import numpy as np
# create Numpy array --> from list

arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr), arr.dtype, arr.shape)


arr2 = np.array([1, 2, 3, 4, 5, "Hello"])
print(arr2, type(arr2), arr2.dtype, arr2.shape)


# 2D arrays --> from 2D list

arr2D = np.array([[1, 2, 3], [4, 5, 6], [7, 8, 3]])
print(arr2D, type(arr2D))
print(arr2D.shape)
```

```python
arr3 = np.zeros((3,3),dtype="int64")
print(arr3)
print(arr3.shape)
arr3[1][2] = 1
print(arr3)
print(arr3.T)
print(arr3.dtype)
```

```python
#indexing and slicing
grid=np.arange(1,10).reshape(3,3)
print(grid)
print(grid[1,2])    #element
print(grid[:,:])    #whole matrix
print(grid[1,:])    #2st row
print(grid[:,1])    #2nd column
print(grid[1:])     #everthing from second row(inclusive)
print(grid[:1])     #everything till second row(exclusive)
print(grid[:,1:])   #from 2nd column
print(grid[:,:1])   #to 2nd column(exlusive)
print(grid[1:,:2])  #bottom from 2nd row and and before 3rd column
print(grid[0,::-1]) #first row in reverse
print(grid[::-1,0]) #first column in reverse
```
